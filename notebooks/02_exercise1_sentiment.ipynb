## 2. Tải mô hình pretrained từ Hub

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
model_id = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSequenceClassification.from_pretrained(model_id)
model.eval()
print(model.config.id2label)
```

## 3. Tokenize một câu mẫu

```python
sentence = "I love this movie. It is amazing!"
inputs = tokenizer(sentence, return_tensors="pt", truncation=True, max_length=512)
print("Tokens:", tokenizer.convert_ids_to_tokens(inputs["input_ids"][0]))
print("Input IDs:", inputs["input_ids"])
print("Attention mask:", inputs["attention_mask"])
```

## 4. Phân tích cảm xúc câu mẫu

```python
with torch.no_grad():
    logits = model(**inputs).logits
probs = torch.softmax(logits, dim=-1)[0]
pred = int(torch.argmax(probs))
print("Sentence:", sentence)
print("Predicted sentiment:", model.config.id2label[pred])
print("Confidence:", round(float(probs[pred]), 4))
```
